```python
import pandas as pd
```

```python
def combine(urlnet_df, htmlphish_df, webphish_df, out):

    # Create unified dataframe
    final_df = pd.DataFrame()

    # Use index as sample number
    final_df["sample"] = range(1, len(webphish_df) + 1)

    # True labels (assumed consistent across datasets)
    final_df["true_label"] = webphish_df["True_Label"]

    # urlnet columns
    final_df["urlnet_label"] = urlnet_df["Predicted_Label"]
    final_df["urlnet_prediction"] = urlnet_df["Malicious_Probability"]

    # htmlphish columns
    final_df["htmlphish_label"] = htmlphish_df["Predicted_Label"]
    final_df["htmlphish_prediction"] = htmlphish_df["Prediction_Probability"]

    # webphish columns
    final_df["webphish_label"] = webphish_df["Predicted_Label"]
    final_df["webphish_prediction"] = webphish_df["Prediction_Probability"]

    # Save final dataframe as Excel
    final_df.to_excel(out, index=False)

    print("✅ Merged file created: merged_results.xlsx")
```

```python
# Load the datasets
urlnet_df = pd.read_excel("/content/drive/MyDrive/Final_Thesis/url_results_d2.xlsx")       # columns: URL, True_Label, Predicted_Label, Malicious_Probability
htmlphish_df = pd.read_excel("/content/drive/MyDrive/Final_Thesis/html_results_d2.xlsx") # columns: HTML, True_Label, Predicted_Label, Prediction_Probability
webphish_df = pd.read_excel("/content/drive/MyDrive/Final_Thesis/combined_results_d2.xlsx")   # columns: URL, HTML, True_Label, Predicted_Label, Prediction_Probability
out = "/content/drive/MyDrive/Final_Thesis/final_d2.xlsx"

combine(urlnet_df, htmlphish_df, webphish_df, out)

# Load the datasets
urlnet_df = pd.read_excel("/content/drive/MyDrive/Final_Thesis/url_results_d3.xlsx")       # columns: URL, True_Label, Predicted_Label, Malicious_Probability
htmlphish_df = pd.read_excel("/content/drive/MyDrive/Final_Thesis/html_results_d3.xlsx") # columns: HTML, True_Label, Predicted_Label, Prediction_Probability
webphish_df = pd.read_excel("/content/drive/MyDrive/Final_Thesis/combined_results_d3.xlsx")   # columns: URL, HTML, True_Label, Predicted_Label, Prediction_Probability
out = "/content/drive/MyDrive/Final_Thesis/final_d3.xlsx"

combine(urlnet_df, htmlphish_df, webphish_df, out)
```
